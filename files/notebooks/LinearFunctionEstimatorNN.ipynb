```python
import numpy as np
```

1. **The Loss Function (MSE)**:
   The Mean Squared Error loss function quantifies the error between the predicted values \( \hat{y}_i \) and the actual values \( y_i \). It is defined as:

   \[
   L = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
   \]

   where:
   - \( n \) is the number of data points.
   - \( y_i \) is the actual value for the \( i \)-th data point.
   - \( \hat{y}_i \) is the predicted value for the \( i \)-th data point, given by the equation \( \hat{y}_i = m x_i + b \).

2. **Substitute \( \hat{y}_i \) in the Loss Function**:
   The predicted value \( \hat{y}_i \) depends on the parameters \( m \) and \( b \):

   \[
   L = \frac{1}{n} \sum_{i=1}^{n} \left( y_i - (m x_i + b) \right)^2
   \]

3. **Partial Derivative with Respect to \( m \) (i.e., \( dm \)):**
   To find the update rule for \( m \), we need to compute the gradient of the loss function with respect to \( m \). The loss is a function of \( m \), so we take the derivative of the loss \( L \) with respect to \( m \).

   Using the chain rule:

   \[
   \frac{\partial L}{\partial m} = \frac{\partial}{\partial m} \left( \frac{1}{n} \sum_{i=1}^{n} (y_i - (m x_i + b))^2 \right)
   \]

   First, apply the chain rule to the squared term \( (y_i - (m x_i + b))^2 \):

   \[
   \frac{\partial}{\partial m} \left( (y_i - (m x_i + b))^2 \right) = 2(y_i - (m x_i + b)) \cdot (-x_i)
   \]

   Simplifying:

   \[
   \frac{\partial L}{\partial m} = \frac{2}{n} \sum_{i=1}^{n} (y_i - (m x_i + b)) (-x_i)
   \]

   This simplifies further to:

   \[
   dm = -\frac{2}{n} \sum_{i=1}^{n} x_i (y_i - (m x_i + b))
   \]

4. **Partial Derivative with Respect to \( b \) (i.e., \( db \)):**
   Similarly, we compute the gradient of the loss function with respect to \( b \).

   \[
   \frac{\partial L}{\partial b} = \frac{\partial}{\partial b} \left( \frac{1}{n} \sum_{i=1}^{n} (y_i - (m x_i + b))^2 \right)
   \]

   Applying the chain rule again:

   \[
   \frac{\partial}{\partial b} \left( (y_i - (m x_i + b))^2 \right) = 2(y_i - (m x_i + b)) \cdot (-1)
   \]

   Simplifying:

   \[
   \frac{\partial L}{\partial b} = \frac{2}{n} \sum_{i=1}^{n} (y_i - (m x_i + b)) (-1)
   \]

   Simplifying further:

   \[
   db = -\frac{2}{n} \sum_{i=1}^{n} (y_i - (m x_i + b))
   \]

5. **Summary of Gradients**:

   - The gradient with respect to \( m \) (i.e., \( dm \)) is:

   \[
   dm = -\frac{2}{n} \sum_{i=1}^{n} x_i (y_i - (m x_i + b))
   \]

   - The gradient with respect to \( b \) (i.e., \( db \)) is:

   \[
   db = -\frac{2}{n} \sum_{i=1}^{n} (y_i - (m x_i + b))
   \]


```python
# Sample data: x and y
x = np.array([1, 2, 3, 4, 5], dtype=float)  # Input features
y = np.array([3, 5, 7, 9, 11], dtype=float)  # Actual values
```

```python
# Initialize parameters m (slope) and b (intercept) randomly
m = np.random.randn()
b = np.random.randn()
```

```python
# Hyperparameters
learning_rate = 0.01  # Step size for parameter updates
epochs = 3000  # Number of iterations

# Number of data points
n = len(x)
```

```python
# Training the neural network
for epoch in range(epochs):
    # Forward pass: calculate the predictions
    y_pred = m * x + b

    # Calculate the Mean Squared Error (MSE) Loss
    loss = (1 / n) * np.sum((y - y_pred) ** 2)

    # Compute gradients (derivatives of loss with respect to m and b)
    # these are effectively the weight and bias of a single layer neuron - a linear function!
    dm = -(2 / n) * np.sum(x * (y - y_pred))  # Gradient with respect to m
    db = -(2 / n) * np.sum(y - y_pred)        # Gradient with respect to b

    # Update parameters using the gradients
    m = m - learning_rate * dm
    b = b - learning_rate * db

    # Print loss every 100 iterations
    if epoch % 1 == 0:
        print(f"Epoch {epoch}: Loss = {loss:.4f}, m = {m:.4f}, b = {b:.4f}")
```

```python
# Output final values for m and b
print(f"\nFinal parameters: m = {m:.4f}, b = {b:.4f}")
```
